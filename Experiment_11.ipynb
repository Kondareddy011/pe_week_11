# Experiment 11: Building a Simple LCEL Chain

Regenerated as a standalone notebook using Google Gemini API.

```python

import os, json, re, getpass
try:
    from google import genai
except ImportError as exc:
    raise ImportError("Install the Gemini SDK first: %pip install -q google-genai") from exc

# Use a new Gemini API key. Input is hidden and not saved in this notebook.
API_KEY = os.getenv("GEMINI_API_KEY") or getpass.getpass("Enter your Gemini API key (input hidden): ")
MODEL = "gemini-2.5-flash"
client = genai.Client(api_key=API_KEY)

def ask_llm(prompt, system="You are a helpful assistant.", temperature=0.2):
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={"system_instruction": system, "temperature": temperature}
    )
    return response.text or ""

print("Gemini client configured. Model:", MODEL)
```

## Experiment Program

```python
try:
    from langchain_core.prompts import ChatPromptTemplate
    from langchain_core.output_parsers import StrOutputParser
    from langchain_google_genai import ChatGoogleGenerativeAI
    prompt = ChatPromptTemplate.from_template("Write a very short story (2-3 sentences) about {topic}.")
    llm = ChatGoogleGenerativeAI(model=MODEL, google_api_key=API_KEY, temperature=0)
    chain = prompt | llm | StrOutputParser()
    for topic in ["a dragon and a knight", "a lost spaceship"]:
        print(topic, "->", chain.invoke({"topic": topic}))
except ImportError:
    print("Install optional dependencies: %pip install -q langchain-core langchain-google-genai")
    for topic in ["a dragon and a knight", "a lost spaceship"]:
        print(topic, "->", ask_llm(f"Write a very short story (2-3 sentences) about {topic}."))
```
